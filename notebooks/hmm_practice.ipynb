# Recreating the HMM used to identify Bacteriophage T4, Recombination and repair protein, C-terminal domain

# STEP 1:
### installing HMMER3
Installed using

 `conda install -c bioconda hmmer`

# STEP 2:
## Downloading seed alignments
Installing seed alignments for the PF21134 PFAM domain
`curl -L "https://www.ebi.ac.uk/interpro/wwwapi/entry/pfam/PF21134/?annotation=alignment:seed&download"  --compressed -o PF21134_seed.sto.gz`

`gunzip PF21134_seed.sto.gz`

# STEP 3:
## Building HMM profile
`hmmbuild test.hmm PF21134_seed.sto`

# STEP 4:
# Searching database
I will use the profile used to search against the uniprot_query_search.fasta. This should identify many of the proteins present in the domain_match.fasta

`hmmsearch test.hmm ../uvsx/data/curated_database/uniprot_query_search.fasta > test.out`

I created a fasta with some sequences that are completely unrelated and some sequences from the uniprot query search.

`hmmsearch test.hmm test_database.fasta > test_2.out`
