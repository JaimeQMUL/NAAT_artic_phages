```python
import pandas as pd
from collections import Counter


classifications = pd.read_csv('fast/fast_result.csv')[['accession','thermal_range']]
temperatures=pd.read_csv('../top_hits/topt/top_hits_tp.csv')[['accession','tOPT']]


import pandas as pd

# Check accession sets are identical
assert set(classifications['accession']) == set(temperatures['accession']), \
    "Accession values do not match"

# Optional: check for duplicates
assert not classifications['accession'].duplicated().any(), \
    "Duplicates in df1"

assert not temperatures['accession'].duplicated().any(), \
    "Duplicates in df2"

# Join classifications and temperatures
joined = pd.merge(
    classifications,
    temperatures,
    on='accession',
    validate='one_to_one'
)


# Fixes

# Removing error
joined = joined[joined['thermal_range'] != 'ERROR']

# Rename one entry of psychrotroph to psychrophile
joined.loc[joined['thermal_range'] == 'psychrotroph', 'thermal_range'] = 'psychrophile'

order = ['psychrophile', 'mesophile', 'thermophile']

joined['thermal_range'] = pd.Categorical(
    joined['thermal_range'],
    categories=order,
    ordered=True
)


# Plot

import matplotlib.pyplot as plt
import numpy as np

data = [
    joined.loc[joined['thermal_range'] == g, 'tOPT']
    for g in order
]

fig, ax = plt.subplots(figsize=(8, 6))

ax.boxplot(data, labels=order)

# scatter overlay
for i, g in enumerate(order, start=1):
    y = joined.loc[joined['thermal_range'] == g, 'tOPT']
    x = np.random.normal(i, 0.04, size=len(y))
    ax.scatter(x, y, alpha=0.7)

ax.set_xlabel('Predicted Thermal Range')
ax.set_ylabel('Predicted Optimal Temperature (celcius)')
ax.set_title('Predicted Optimal Temperature by Predicted Thermal Range')

plt.tight_layout()
plt.show()

# Count occurrences
counts = Counter(joined['thermal_range'])

# Plot pie chart
plt.pie(
    counts.values(),
    labels=counts.keys(),
    autopct='%1.1f%%'
)

plt.axis('equal')
plt.show()
```

```python
import pandas as pd
from collections import Counter


classifications = pd.read_csv('summary/summary_result.csv')[['accession','thermal_range']]
temperatures=pd.read_csv('../top_hits/topt/top_hits_tp.csv')[['accession','tOPT']]


import pandas as pd

# Check accession sets are identical
assert set(classifications['accession']) == set(temperatures['accession']), \
    "Accession values do not match"

# Optional: check for duplicates
assert not classifications['accession'].duplicated().any(), \
    "Duplicates in df1"

assert not temperatures['accession'].duplicated().any(), \
    "Duplicates in df2"

# Join classifications and temperatures
joined = pd.merge(
    classifications,
    temperatures,
    on='accession',
    validate='one_to_one'
)


# Fixes

# Removing error
joined = joined[joined['thermal_range'] != 'ERROR']

# Rename one entry of psychrotroph to psychrophile
joined.loc[joined['thermal_range'] == 'psychrotroph', 'thermal_range'] = 'psychrophile'

order = ['psychrophile', 'mesophile', 'thermophile']

joined['thermal_range'] = pd.Categorical(
    joined['thermal_range'],
    categories=order,
    ordered=True
)


# Plot

import matplotlib.pyplot as plt
import numpy as np

data = [
    joined.loc[joined['thermal_range'] == g, 'tOPT']
    for g in order
]


colors = ['blue', 'orange', 'red']  # match your categories
fig, ax = plt.subplots(figsize=(8, 6))
bp = ax.boxplot(data, labels=order, patch_artist=True, showmeans=True)

# Colour each box
for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.4)

# Colour medians
for median, color in zip(bp['medians'], colors):
    median.set_color(color)

# Colour whiskers + caps (need grouping: 2 per box)
for i, color in enumerate(colors):
    bp['whiskers'][2*i].set_color(color)
    bp['whiskers'][2*i+1].set_color(color)
    bp['caps'][2*i].set_color(color)
    bp['caps'][2*i+1].set_color(color)

# Mean lines (if showmeans=True)
for mean, color in zip(bp['means'], colors):
    mean.set_marker('o')
    mean.set_markerfacecolor(color)
    mean.set_markeredgecolor(color)

# Scatter overlay (match same colours)
for i, (g, color) in enumerate(zip(order, colors), start=1):
    y = joined.loc[joined['thermal_range'] == g, 'tOPT']
    x = np.random.normal(i, 0.04, size=len(y))
    ax.scatter(x, y, alpha=0.6, color=color)

ax.set_xlabel('Predicted Thermal Range')
ax.set_ylabel('Predicted Optimal Temperature (°C)')
ax.set_title('Predicted Optimal Temperature by Predicted Thermal Range')
plt.tight_layout()

plt.savefig('uvsx/plots/tmp-seq2opt_boxplot.png', bbox_inches='tight', dpi=300)
plt.show()

# Count occurrences
counts = Counter(joined['thermal_range'])

colours=['orange', 'blue', 'red']

fig, ax = plt.subplots(figsize=(6, 6))

wedges, texts, autotexts = ax.pie(
    counts.values(),
    autopct='%1.1f%%',
    startangle=0,
    colors=colours,
    pctdistance=1.1,
)
# Move the small-slice percentages
autotexts[1].set_position((1.1, 0))
autotexts[2].set_position((1.1, -0.1))

ax.legend(
    wedges,
    counts.keys(),
    title="Thermal range",
    loc="lower right",
)
ax.set_title('Breakdown of Classifications from TMP')

ax.axis('equal')
plt.savefig('uvsx/plots/tmp_class_pie.png')
plt.show()
```

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
from src.seq2topt.functions import *
import timeit
import torch
from src.seq2topt.model import MultiAttModel
import esm
from src.biotools.fasta_tools import *
from src.seq2topt import *

# LOAD tOPT MODEL
# path to model weights
topt_pth = '../../src/seq2topt/model_topt_window.3_r2.0.57.pth';# you can download the weights from release v1.0.0
# define device
device = torch.device('cpu')
# define seq2Topt model
emb_dim= 320; window=3; n_head = 4; n_RD = 4;
model = MultiAttModel( emb_dim, window, n_head, n_RD)
model.to(device);
# load weights into the model
model.load_state_dict(torch.load( topt_pth, map_location=device  ));
model.eval();

# LOAD ESM-2
# We use esm2_t6_8M_UR50D, 6 layers, dimension size = 320.
esm2_model, alphabet = esm.pretrained.esm2_t6_8M_UR50D() # 6 layers
esm2_model = esm2_model.to(device)
esm2_batch_converter = alphabet.get_batch_converter()

def predict(seq):
    # seq: input protein sequence
    inputs = [('Temp', seq)]
    batch_labels, batch_strs, batch_tokens = esm2_batch_converter(inputs)
    batch_tokens = batch_tokens.to(device=device, non_blocking=True)
    with torch.no_grad():
        emb = esm2_model(batch_tokens, repr_layers=[6], return_contacts=False)
    emb = emb["representations"][6]
    emb = emb.transpose(1,2)
    emb = emb.to(device)
    with torch.no_grad():
        preds = model( emb )
    return preds

topt_max=120

database='../../1-DatabaseCuration/filtered/sequences/filtered_curated_database.fasta'

uvsxt=ExtractSequence('7Z3M', database)
temp_t=(predict(uvsxt)*topt_max)

uvsxp=ExtractSequence('9GBG', database)
temp_p=(predict(uvsxp)*topt_max)

uvsx=ExtractSequence('P04529', database)
temp_u=(predict(uvsx)*topt_max)
```
